### Скрипт на Python

Скрипт - обычный текстовый файл, содержащий код. Сохраним строку с программой приветствия в файл `hello_world.py`

Расширение `.py` не является обязательным для запуска программы, но оно позволяет *Python* воспринять программу как *модуль* - способ переиспользовать код между программами.

В *Python* рекомендуется все названия писать строчными буквами, а слова в названии разделять подчеркиванием.

У этого стиля есть название - *snake case*.

Но никто не мешает назвать нам скрипт `HelloWorld.py` или `helloWorld.py` - зависит от ваших предпочтений.

Рекомендации по стилю содержатся в специальном документе - [PEP-8](https://peps.python.org/pep-0008/), мы будем следовать ему.

Скрипт можно вызвать командой:

````python
!python hello_world.py
````

Простые программы можно передать интерпретатору Python в качестве *параметра*:

````python
!python -c 'print("Hello, World!")'
````

В скрипте каждая новая строка означает новую операцию для интерпретатора:

````python
print("Hello, World!")
print("Hello, students!")
````

В скрипте это скорее мешает читаемости, но вполне подойдет для вызова многострочной программы через командную строку:

## Организация кода

Весь код в *Python* можно разбить на *инструкции* (*statements*) - блоки кода, отвечающие за определенные виды операций в интерпретаторы.

Некоторые инструкции мы называем *выражениями* (*expressions*) - это операции, по результатам которой мы получаем новое *значение*.

С *выражениями* мы уже немного знакомы:

````python
(1 + (2 * (3 // (4 - (5 ** 6)))))
````

Скобками выделены выражения, а их результатами являются другие значения, которые в свою очередь подставляются в другие выражения.

Вызов функции также является выражением, ведь ее результат - значение:

````python
10 + int("20")
````

### Инструкции

#### Переменные

````python
simple_pi = 22/7
````

Почему же это не выражение?

Значением в коде выше будет `22/7`, присваивание связывает это значение с именем, но не создает его.

Давайте рассмотрим на другом примере:

````python
a = 100_000
b = a

b == a
````

Как мы видим, значение обеих переменных равно.

Но на самом деле Python работает немного по-другому: он создал объект - число `100_000`, присвоил его переменной `a`, а затем его же присвоил переменной `b`.

То есть и за `a`, и за `b` скрывается одна и та же ячейка внутри памяти *Python*!

Мы можем убедиться в этом с помощью оператора `is`:

````python
a is b
````

Заведем еще одну переменную:

````python
c = 100_000
a == c
````

Но тот же это объект в *Python*?

````python
a is c
````

Нет, ведь мы явно создали новый объект, пусть и с тем же значением, что уже существует.

В *Python* есть служебная функция `id()`, которая указывает идентификатор - некое число, под которым существует значение внутри его памяти:

````python
print(id(a))
print(id(b))
print(id(c))
````

Для чисел и строк это свойство переменных не особо важно: мы не можем изменить значение объекта, только создать новый:

````python
a = a * 10
print(a)
print(b)
````

Такие типы данных называются *неизменяемыми* (*immutable*).

Но есть и изменяемые типы данных (которые мы рассмотрим позже), для которых будет важно помнить - переменные могут ссылаться на один и тот же объект.

#### import

Выше мы устанавливали внешние библиотеки и загружали их в действующую сессию при помощи инструкции `import`.

Но в *Python* уже есть огромное количество библиотек, поставляемых сразу вместе с интерпретатором, например библиотека для математики `math`:

````python
import math

math.pi
````

`import` загружает все, что хранится в модуле, в специальном *пространстве имен*, чтобы не допустить конфликтов с существующими переменными.

Поэтому значения внутри модуля будут доступны как `имя_модуля.имя_переменной`:

````python
math.sqrt(4)
````

Мы можем загрузить только то, что нам нужно из модуля, при помощи инструкции `from ... import ...`:

````python
from math import ceil, floor

print(ceil(1.5))
print(floor(1.5))
````

Данные функции нужны для округления дробного числа в большую или меньшую сторону соответственно.

Что будет, если импортировать переменную, которая раньше существовала?

#### del

Иногда нам нужно удалить какую-то переменную из памяти:

````python
del a
a
````

Представим, что мы присвоили одно значение трем переменным:

````python
a = 1_000_000
b = a
c = a
````

Что можно записать более коротким способом:

````python
a = b = c = 1_000_000
````

Когда мы удалим `a`, `b` и `c`, у объекта, который стоит за ними, числа `1_000_000`, больше не будет присвоенных переменных, и *Python* удалит его из своей памяти.

````python
del a
del b
del c
````

В этом и заключается основное назначение `del`, в обычных же случаях нам не нужно явно удалять какую-либо переменную.

#### if/elif/else

Предыдущие инструкции состояли из одной строки, перейдем к первой сложной инструкции.

Код обычно "ветвится" в зависимости от значений на каком-то этапе исполнения, для этого у нас есть инструкция `if`:

````python
a = 45

if a % 2 == 0:
    print("Число четное")

print("Число", 45)
````

Блок `if` проверяет логическое значение, и если оно `True` - выполняет блок кода.

Заметьте, что блок, который выполнится при условии, отделен от обычного кода 4 пробелами (или одним табом) - в *Python* это важно!

Так интерпретатор понимает, когда блок начинается и заканчивается.

Каждый `if` не зависит друг от друга:

````python
a = 6

if a % 2 == 0:
    print("Число делится на 2")

if a % 3 == 0:
    print("Число делится на 3")

if a % 6 == 0:
    print("Число делится на 6")
````

Но что, если мы хотим связать условия?

Скажем, написать программу, которая проверит, делится ли программа на 2, 3 или сразу на 6?

Для этого у нас есть блок `elif`:

````python
a = 6

if a % 6 == 0:
    print("Число делится на 6")
elif a % 2 == 0:
    print("Число делится на 2")
elif a % 3 == 0:
    print("Число делится на 3")
````

Если мы хотим представить условие как "во всех остальных случаях" - используем блок `else`:

````python
a = 83

if a % 6 == 0:
    print("Число делится на 6")
elif a % 2 == 0:
    print("Число делится на 2")
elif a % 3 == 0:
    print("Число делится на 3")
else:
    print("Число не делится ни на одно из чисел")
````

Единственный блок, который обязателен - `if`, остальные по нашему усмотрению.

Блоки `if` могут быть вложены друг в друга.

Представим, что мы вычисляем скидку для клиента:

````python
purchase_amount = 1200
premium_member = True
````

Мы можем выразить условия вычисления так:

````python
if purchase_amount >= 1000 and premium_member == True:
    print("Применяем скидку 25%")
elif purchase_amount >= 1000 and premium_member == False:
    print("Применяем скидку 10%")
else:
    print("Нет скидки")
````

Либо упростить условия:

````python
if purchase_amount >= 1000:
    if premium_member == True:
        print("Применяем скидку 25%")
    else:
        print("Применяем скидку 10%")
else:
    print("Нет скидки")
````

Если значение уже `True` либо `False`, его можно сразу подставить в `if` блок:

````python
if premium_member:
    print("Премиальный пользователь")
````

Наконец, вспомним про неявное преобразование значений в `bool` в *Python*.


Проверку на четность можно записать так:

````python
a = 4

if a % 2:
    print("Число нечетное")
else:
    print("Число четное")
````
