# Коллекции в Python

Ранее мы с вами работали с простыми типами (числа, boolean) и простыми последовательностями (строками, состоящими из строк). Рассмотрим типы данных, которые могут включать в себя другие типы данных - *коллекции*.

## Кортежи

*Кортеж* (`tuple`) — упорядоченная коллекция, которую нельзя изменить после создания.

Он удобен для данных с фиксированной структурой: координат, даты, записи о сотруднике.

Кортеж записывают через запятые. Скобки улучшают читаемость, но именно запятая создаёт кортеж.


```python
point = (27.56, 53.90)
# можно записать как
point = 27.56, 53.90
print(point)
print(type(point))
```

У кортежа из одного элемента обязательно должна быть завершающая запятая.

```python
one_item = (42,)
print(one_item)
print(type(one_item))
```

Просто скобок недостаточно:

```python
just_int = (42)
type(42)
```

Минимальный кортеж - пустой - записывается просто как скобки:

```python
empty = ()
print(empty)
print(type(empty))
```

Кортежи - последовательности. У них есть длина:

```python
print(len(point))
print(len(one_item))
print(len(empty))
```

Кстати, что будет, если преобразовать в `bool` пустой кортеж?

```python
bool(empty)
```

Можем получить элементы по индексу или срезу:

```python
print(point[0])
print(point[::-1])
```

Можно искать элемент в кортеже:

```python
point = (27.56, 53.90)
print(27 in point)
print(point.index(27.56))
print(point.index(27))
```

И посчитать количество элементов:

```python
point.count(27.56)
```

### Распаковка

Кортеж задуман так, что порядок элемента подразумевает его назначение.

Возьмем `point`:

```python
point
```

Это пара координат *(x, y)* или *(latitude, longitude)*. Мы можем получить каждый элемент по индексу:

```python
x = point[0]
y = point[1]
print(f"{x=} {y=}")
```

В Python есть более удобный способ определить переменные из последовательности - распаковка:

```python
x, y = point
print(f"{x=} {y=}")
```

Переменных должно быть столько, сколько элементов в последовательности, иначе будет ошибка:

```python
x, y, z = point
```

Распаковка доступна для **всех** последовательностей:

```python
a, z = "az"
print(f"{a=} {z=}")
```

Вернемся к кортежам. Кортежи могут хранить в себе любое количество любых типов данных:

```python
employee = (101, "Анна", "аналитик")
```

Этот кортеж - запись личного номера, имени и роли сотрудника:

```python
employee = (101, "Анна", "аналитик")
employee_id, name, role = employee
print(f"Номер сотрудника - {employee_id}, имя - {name}, роль - {role}")
```

Иногда нам нужны не все элементы кортежа. Тогда оставшиеся элементы можно отправить в переменную с `*` в начале (это будет уже не кортеж, а *список*):

```python
employee_id, *rest = employee
print(employee_id)
print(rest)
print(type(rest))
```

Так можно получить переменные с начала и/или конца последовательности:

```python
route = ("Минск", "Вильнюс", "Рига", "Таллин")

start, *stops = route
print("Старт:", start)
print("Остальные остановки:", stops)

*stops, finish = route
print("Финиш:", finish)
print("Остальные остановки:", stops)

start, *stops, finish = route
print("Старт:", start)
print("Финиш:", finish)
print("Остальные остановки:", stops)
```

Мы уже видели распаковки раньше, например:

```python
for index, element in enumerate("abcde"):
    print(index)
    print(element)
```

На самом деле это кортеж!

```python
for t in enumerate("abcde"):
    print(t)
    print(type(t))
```

Кортеж может содержать любые типы, в том числе и сами кортежи:

```python
city = ("Минск", "Беларусь", True, (53.9, 27.57))
```

Можем распаковать все значения в два подхода:

```python
name, country, is_capital, coordinates = city
latitude, longitude = coordinates
print(f"{name=} {country=} {is_capital=} {latitude=} {longitude=}")
```

Или сразу - вложенной распаковкой:

```python
(name, country, is_capital, (latitude, longitude)) = city
print(f"{name=} {country=} {is_capital=} {latitude=} {longitude=}")
```

## Списки



**Список** (`list`) — упорядоченная изменяемая коллекция. Он подходит, когда элементы нужно добавлять, удалять, заменять или сортировать.

Мы уже создавали списки из строки методом `.split()`:

```python
sentence = "Привет всем учащимся"
words = sentence.split()
print(words)
print(type(words))
```

Но элементы списка не ограничиваются строками.

Элементы списка, как и кортежа, могут быть разных типов, но на практике обычно хранят однотипные данные.

```python
products = ["хлеб", "молоко", "яблоки"]
prices = [2, 3, 5]
empty = []
```

Списки поддерживают все операции последовательностей:

```python
print(prices[0])
print(prices[1:])
print(prices[::-1])

for price in prices:
    print(price)

print(len(prices))

print(4 in prices)
print(prices.count(2))
print(prices.index(3))

first_price, second_price, *rest = prices
print(first_price, second_price, rest)
```

Пустой список - также False, как и пустой кортеж:

```python
bool(empty)
```

В чем же отличие от кортежей?

Списки - первый *изменяемый (mutable)* тип, с которым мы знакомимся.

```python
print(id(prices))
prices[0] = 1.5
print(prices)
print(id(prices))
```

Сочетание `.append()` и `.pop()` дает нам возможность реализовать такую структуру данных, как *стек (stack)* - "последнее на вход - первое на выход".

Рассмотрим задачу. Нам нужно удалить пары символов в строке:

```python
string = "abbaca"
stack = []
for char in string:
    print(stack)
    print(char)
    if len(stack) == 0:  # если символов нет - добавляем
        stack.append(char)
    elif stack[-1] == char:  # если последний добавленный символ такой же, как и текущий - дубликат, удаляем
        stack.pop()
    else:  # если не дубликат - тоже добавляем
        stack.append(char)
    print(stack)
```

Решим задачу при помощи стека. Проверим, верно ли расставлены скобки в строке:
* `"((())())"` - верно
* `"(()"` - неверно
* `")("` - неверно
* `"(()))("` - неверно
* `"(())()"` - верно

```python
string = input("Введите скобки: ")
stack = []
for char in string:
    if not stack and char == ")":
        print("Неверно")
        break
    if char == "(":
        stack.append("(")
    else:
        stack.pop()

if not stack:
    print("Верно")
else:
    print("Неверно")
```

 ### Сортировка и копирование

Метод `.sort()` меняет исходный список, упорядочив по возрастанию:

```python
prices = [12.5, 7.0, 20.0, 9.5]
print(id(prices))
prices.sort()
print(prices)
print(id(prices))
```

Или по убыванию:

```python
prices.sort(reverse=True)
print(prices)
print(id(prices))
```

Но часто нам надо сохранить и оригинальный список, и получить отсортированную версию. В этом нам поможет функция `sorted()`:

```python
prices = [12.5, 7.0, 20.0, 9.5]
print(id(prices))
ascending = sorted(prices)
print(ascending)
print(id(ascending))
descending = sorted(prices, reverse=True)
print(descending)
print(id(descending))
```

То, что список - изменяемый тип, важно помнить: это может привести нас к ошибкам!

Присваивание `copy = original` не копирует список: обе переменные указывают на один объект.

```python
original = [1, 2, 3]
copy = original

original.sort(reverse=True)
copy.append(99)

print("original:", original)
print("copy:", copy)
```

В итоге меняя один и тот же список в двух разных местах, мы можем получить не те данные, которые ожидаем.

Для того, чтобы действительно скопировать список, можно воспользоваться слайсами:

```python
original = [1, 2, 3]
print(id(original))
copy = original[:]
print(id(copy))

original.sort(reverse=True)
copy.append(99)

print("original:", original)
print("copy:", copy)
```

Или методом `.copy()`:

```python
original = [1, 2, 3]
print(id(original))
copy = original.copy()
print(id(copy))

original.sort(reverse=True)
copy.append(99)

print("original:", original)
print("copy:", copy)
```

### List comprehensions


*List comprehension* — компактный способ построить новый список из существующей
последовательности. Читаем конструкцию слева направо: «что добавить — откуда взять — при каком условии».

> [выражение for элемент in коллекция if условие]

Вспомним задачу про получение квадратов чисел:

```python
squares = []

for i in range(1, 11):
    squares.append(i ** 2)

print(squares)
```

Вместо этого можно просто создать список при помощи *list comprehension*:

```python
squares = [i ** 2 for i in range(1, 11)]
print(squares)
```

Можем отфильтровать данные для списка:

```python
temperatures = [18, 23, 27, 16, 25]
warm_days = [temperature for temperature in temperatures if temperature >= 23]
print(warm_days)

words = ["дом", "программирование", "кот", "коллекция"]
long_words = [word for word in words if len(word) > 5]
print(long_words)
```

А также совместить преобразование и фильтрацию в одном выражении.

Вместо:

```python
prices = [5, 12, 30, 8, 20]

discounted = []

for price in prices:
    if price >= 10:
        discounted.append(price * 0.9)
print(discounted)
```

Записать:

```python
prices = [5, 12, 30, 8, 20]
discounted = [price * 0.9 for price in prices if price >= 10]
print(discounted)
```

```python
[price if price < 10 else price * 0.9 for price in prices]
```

В Python есть метод `.join()`, который позволяет превратить последовательность опять в строку.

Например:

```python
"-".join("abc")
```

где `"-"` - разделитель, и каждая буква из строки будет объединена этим разделителем.

Можно считать этот метод обратным для метода `.split()`:

```python
s = "один два три"
words = s.split()
joined = " ".join(words)
print(joined)
print(s == joined)
```

Так вот, что если мы хотим перечислить цены из списка `discounted` через запятую?

```python
",".join(discounted)
```

Каждый элемент из последовательности, которая передается в `.join()`, должен быть строкой!

Преобразуем тип при помощи list comprehension:

```python
discounted_strings = [str(price) for price in discounted]
print(",".join(discounted_strings))
```

Наконец, в list comprehensions могут быть вложенные циклы.

Выведем таблицу умножение от 1 до 5 включительно, циклом:

```python
products = []

for i in range(1, 6):
    for j in range(1, 6):
        product = i * j
        products.append(product)
        print(f"{i}*{j}={product}")
```

Или list comprehension:

```python
[i * j for i in range(1, 6) for j in range(1, 6)]
```

Когда лучше обычный цикл?

Если действие сложное, содержит несколько шагов или побочные эффекты (например, `input()` или `print()`), обычный цикл читается легче.

List comprehension прежде всего строит **новый список**.

## Словари

*Словарь* (`dict`) хранит пары `ключ: значение`. В отличие от кортежа, ключ - не позиция значения, а некое имя.


```python
person = {
    "name": "Ирина",
    "age": 34,
    "city": "Минск",
}
print(person)
```

```python
person = ("Ирина", 34, "Минск")
```

Пустой словарь (естественно `False`):

```python
empty = {}
bool(empty)
```

Ключи словаря уникальны:

```python
person = {
    "name": "Ирина",
    "age": 34,
    "age": 40,
    "city": "Минск",
}
print(person)
```

Ключами могут быть любые значения, для которых работает функция `hash()` (то есть неизменяемые типы данных):

```python
hash(1)
hash(1.5)
hash(True)
hash("abc")
hash((1, 2, 3))
hash([1, 2, 3])
```

```python
{0: "A", 1: "B", 2: "C"}
```

Мы можем получить значение по ключу, что немного напоминает индексы в последовательностях:

```python

person["name"]
```

Словари изменяемы, значит мы можем поменять значение или добавить новую пару `ключ: значение`:

```python

person["email"] = "irina@example.com"   # новый ключ
person["city"] = "Брест"               # изменение значения
print(person)
```

Что, если ключа нет?

```python
person["status"]
```

`dictionary[key]` вызывает `KeyError`, если ключа нет.

```python
settings = {"theme": "dark", "language": "ru"}

print(settings.get("theme"))
print(settings.get("font_size"))
print(settings.get("font_size", 14))
print("language" in settings)
```

`None` - специальное значение, единственный представитель типа `NoneType`. Буквально "ничего" в Python.

Словарь - не последовательность, ведь у него нет индексов. Но есть длина:


```python
prices = {"кофе": 8.5, "чай": 5.0, "какао": 7.5}
print(len(prices))
```

### Dict comprehensions

По аналогии со списками вместо цикла мы можем создать словарь на лету:

> {ключ:значение for элемент in коллекция if условие}

Например, число и его квадрат. Вместо цикла:

```python
d = {}
for i in range(1, 5):
    d[i] = i ** 2
print(d)
```

Удобнее:

```python
{i: i ** 2 for i in range(1, 5)}
```

Можем получить словарь, где ключ - индекс, а значение - элемент последовательности, при помощи `enumerate()`:

```python
{i: word for i, word in enumerate(["яблоко", "капуста", "банан"])}
```

Оно же

```python
dict(enumerate(["яблоко", "капуста", "банан"]))
```

Или объединив две последовательности через `zip()`:

```python
names = ["Jack", "John", "Jim"]
ages = [18, 20, 19]

d = {name: age for name, age in zip(names, ages)}
print(d)
```

Оно же

```python
dict(zip(names, ages))
```

Уберем неподходящие ключи, перебрав пары `ключ: значение` с `.items()`:

```python
{name: age for name, age in d.items() if name.endswith("k")}
```

Или неподходящие значения:

```python
{name: age for name, age in d.items() if age <= 18}
```

Напишите программу, которая принимает список слов и возвращает словарь, где ключами являются слова, а значениями — то, сколько раз они встретились в списке.

```python
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
word_counts = {}

for word in words:
    word_counts[word] = words.count(word)

print(word_counts)
```

У вас есть два списка покупок с ценами. Объедините их в один словарь. Если товар есть в обоих списках, запишите большую цену.

```python
shop_a = {"хлеб": 2, "молоко": 3}
shop_b = {"молоко": 4, "сыр": 10}
merged_shop = {}

for key in shop_a:
    if key not in shop_b:
        merged_shop[key] = shop_a[key]
    elif shop_b[key] > shop_a[key]:
        merged_shop[key] = shop_b[key]
    else:
        merged_shop[key] = shop_a[key]
for key in shop_b:
    if key not in merged_shop:
        merged_shop[key] = shop_b[key]


print(merged_shop)
```

Дан словарь, где ключи — имена студентов, а значения — их средний балл. Найдите и выведите имя студента с самым высоким баллом.

```python
grades = {"Алексей": 8.5, "Мария": 9.8, "Дмитрий": 7.4}

max_score = 0
max_score_name = ""

# Подсказка: можно использовать итерацию по парам .items() и обновлять переменные лучшего студента и его балла
for student, score in grades.items():
    if score > max_score:
        max_score = score
        max_score_name = student

print(max_score_name)
```

Напишите программу, которая принимает список слов и группирует их в словарь по первой букве. Ключом должна быть буква, а значением — список слов, начинающихся на эту букву.

```python
words = ["apple", "banana", "apricot", "cherry", "blueberry"]
grouped_words: dict[str, list[str]] = {}

for word in words:
    print(grouped_words)
    first_letter: str = word[0]
    if first_letter not in grouped_words:
        grouped_words[first_letter] = []
    groups: list[str] = grouped_words[first_letter]
    groups.append(word)

print(grouped_words)
```

У вас есть словарь, представляющий склад товаров. Ключ — название товара, значение — другой словарь с ценой (`price`) и количеством (`quantity`). Посчитайте общую стоимость всего склада.

## Множества

*Множество* (`set`) - коллекция, созданная для хранения строго **уникальных** элементов.

Вы можете представить множество как словарь без значения, что выражается даже в синтаксисе:

```python
{"a", "b", "c"}
```

Пустой `set` создается через `set()`, ведь `{}` уже используется пустым словарем:

```python
d = {}
s = set()
print(type(d))
print(type(s))
print(bool(s))
```

Как у словарей ключи должны быть хэшируемыми, так и у множеств элементы. То есть только неизменяемые элементы!

```python
{1, "a", True, None, (1, 2, 3)}
{[4, 5, 6]}  # Будет ошибка
```

Все элементы уникальны, значит повторы игнорируются:

```python
{1, 2, 3, 1, 2, 3, 1, 2, 3}
```

Можем получить длину:

```python
len({1, 2, 3})
```

Пройтись по элементам в цикле:

```python
for unique_word in {"стол", "стул", "ложка", "стул"}:
    print(unique_word)
```

Но это не последовательность! Индексы не поддерживаются, значит и элементы не упорядочены:

```python
s = {1, 2,3}
s[0]
```

Можем быстро получить уникальные элементы последовательности - просто превратив в `set()`:

```python
set("abracadabra")
```

Проверка, есть ли элемент в множестве, достаточно быстрая:

```python
1 in {4, 5, 6, 5}
```

Можем найти *пересечение* множеств (все элементы из обеих множеств):

```python
{1, 2, 3} | {2, 3, 4}
```

Или их *объединение* (все пересечения из обеих множеств):

```python
{1, 2, 3} & {2, 3, 4}
```

Оставим все элементы из левого множества, которых нет в правом:

```python
{1, 2, 3} - {2, 3, 4}
```

Или все элементы из обеих множеств, которые не пересекаются:

```python
{1, 2, 3} ^ {2, 3, 4}
```

Входит ли одно множество в другое?

```python
print({1, 2, 3} > {2, 3, 4})
print({1, 2, 3} > {2, 3})
```

Множества изменяемы, значит мы можем добавлять (`.add()`) и удалять (`.discard()`) элементы:

```python
s = {1, 2, 3}
s.add(3)  # не будет ошибки
s.add(4)
s.discard(3)
s.discard(5)  # не будет ошибки
```

Если нам нужен неизменяемое множество, можем использовать `frozenset()`:

```python
s = frozenset("abc")
print(s)
s.add("d")
```

Обратите внимание - мы передаем в него последовательность!

Такое множество может быть полезно как ключ словаря:

```python
{
    "a": 1,
    "b": ["a", "b", "c"],
    frozenset("cde"): False
}
```
