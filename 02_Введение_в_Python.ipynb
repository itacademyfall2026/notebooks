```python
print("Hello, World!")
```

## Вывод данных

В интерпретаторе на экран выводится любое значение:

```python
"Hello!"
```

А также результат выражения:

```python
2 * 2
```

Во время исполнения программы вне интерпретатора нам нужно выводить данные на экран. Базовая функция - `print()`

```python
print(10 + 20)
```

В эту функцию можно передать несколько значений - она отобразит их, соединив пробелом:

```python
print("a", 10, True)
```

Ну или тем символом, который мы захотим использовать - при помощи **параметра** *sep*:

```python
print("abc", "def", sep="_")
```

### Особенности Python

*   Всё в Python - **объекты**
*   Каждый объект - **значение** определенного **типа**
*   Каждый тип поддерживает ограниченный набор **операций**
*   Большинство операций подразумевают **явно** конкретные типы
*   Некоторые операции поддерживают **неявное** преобразование типов (в основном чисел)





## Типы в Python

### Строки



Это тип для представления текстовой информации.

Создается при помощи кавычек:

*   одинарных

```python
'Hello, World!'
```

*   двойных



```python
"Hello, World!"
```

*   три пары кавычек - для многострочного текста



```python
"""Змейка плавно код читает,
Отступы важны здесь всем.
Каждый строчки понимает
Смысл без лишних сложных схем.
"""
```

Минимальная строка - без символов ('пустая' строка)

```python
''
```

Строки могут содержать 'экранируемые' символы - при помощи знака `\`

Например, чтобы создать строку с апострофом:

```python
'Child\'s play'
```

Или отображения переноса строки:

```python
"abc\ndef"
```

Строки могут иметь разные кодировки, по умолчанию - Unicode.

```python
"\U0001F602"
```

Любой тип можно преобразовать в строку при помощи функции `str()`

```python
str(10)
```

## Переменные

Смысл переменной - переиспользовать какое-то значение дальше в коде.

```python
a = 10
a + 20
```

Тип переменной не ограничен:

```python
a = 3.5
a = "abc"
a = 10
a
```

Можно присвоить значение одной переменной другой переменной:

```python
b = a
b
```

Можно изменить значение переменной:

```python
b = b + 10
b
```

И поменять значение переменных местами:

```python
a, b = b, a
```

```python
a
```

```python
b
```

Переменной *можно* (но необязательно) указать тип:

```python
a: int = 10
```

При этом не происходит проверка на соответствие типа переменной и значения:

```python
a: int = "abc"
```

Переменную можно объявить заранее:

```python
x: float
```

Но ее нельзя использовать - у нее нет значения!

```python
print(x)
```
