# Алгоритмы

Алгоритм - способ решения задачи, который мы в дальнейшем реализуем в виде кода.

Чаще всего нам не нужно придумавать алгоритмы с нуля, мы используем уже существующие подходы или комбинируем их.

Умение выбрать наиболее подходящий с точки зрения эффективности алгоритм не менее важно, чем умение писать чистый код.

## Простейшие алгоритмы

Мы уже знакомы с рядом простых алгоритмов, попробуем их формализовать для всех задач.

### Накопитель

Это алгоритм, который собирает в цикле результат обработки последовательности в специальной переменной.

Например, сумма:

```python
nums = [3, 4, 5, 6]

total = 0

for num in nums:
    total += num

print(total)
```

Он же

```python
sum(nums)
```

Или произведение

```python
total = 1

for num in nums:
    total *= num

print(total)
```

Он же

```python
from math import prod

prod(nums)
```

### Счетчик

Похоже на накопитель, но мы накапливаем не значения элементов, а количество тех элементов, что подходят нашему условию.

Например, четные числа:

```python
count = 0

for i in range(100):
    if not (i % 2):
        count += 1
print(count)
```

### Минимум/максимум

Ищем крайнее значение, которое подходит под наше условие.

В специальной переменной храним текущий минимум/максимум и сравниваем с каждым новым элементом:

```python
numbers = [7, 2, 9, 4, 6]

maximum = numbers[0]

for number in numbers[1:]:
    if number > maximum:
        maximum = number

print(maximum)
```

В Python есть встроенные функции `min()` и `max()`, которые работают как для коллекций:

```python
print(min(numbers))
print(max(numbers))
```

Так и для отдельных значений:

```python
min(4, 3, 5)
```

### Фильтрация

Мы выберем только те элементы, которые подходят под условия.

Обычно для этих целей создается новая коллекция, куда мы собираем подходящие элементы.

Например, выберем положительные числа:

```python
numbers = [-2, 5, 0, 7, -1, 3]

result = []

for number in numbers:
    if number > 0:
        result.append(number)

print(result)
```

## Сложность алгоритмов

Определив тип алгоритма, мы можем примерно представить, сколько времени он отработает в зависимости от входных данных - *сложность по времени*.

### Линейная сложность

Это случай, когда время работы алгоритма растет примерно пропорционально количеству данных.

Чтобы накопить значение, посчитать элементы, найти минимум/максимум или отфильтровать подходящие, придется пройти по всем элементам коллекции и провести нужную операцию с каждым из них.

А что насчет поиска?

Найдем число 7 в списке:

```python
numbers = [4, 9, 2, 7, 5]

for number in numbers:
    if number == 7:
        print("Нашли!")
        break
```

Элемент может быть в любом месте списка, и в лучшем случае он будет в самом начале списка, а алгоритм отработает за 1 шаг. Но для оценки алгоритмов мы должны ориентироваться на **худший** случай.

Пример выше очень прост - в списке 5 элементов, в худшем случае алгоритм отработает за 5 шагов. Но настоящий алгоритм абстрагируется от конкретных данных, значит мы должны придумать абстрактный способ оценки.

### Big O Notation

Это способ записи сложности алгоритмов, в котором количество элементов представлено как переменная.

В случае линейной сложности мы записываем **O(n)**, где *n* - количество элементов, которые надо обработать. В Python для последовательности `s` это `len(s)`.

Значит ли это, что мы выполним ровно *n* операций? Не обязательно, вспомним про поиск - ориентируемся на худший случай.

Также мы учитываем только те операции, которые растут с количеством данных. Например, вспомним задачу про поиск палиндрома:

```python
s = "А роза упала на лапу Азора!"

letters = ""

for char in s:
    if char.isalpha():
        letters += char.lower()

if letters == letters[::-1]:
    print("Палиндром!")
```

Сначала мы перебрали все элементы и оставили только буквы в нижнем регистре. Затем отобразили эти буквы (то есть по сути перебрали буквы в обратном порядке) и сравнили два результата. Получается, сложность алгоритма *O(2n)*?

При оценке сложности алгоритмов мы отбрасываем постоянные коэффициенты или дополнительные операции. И для пустой строки, и для строки на миллион символов алгоритм дважды переберет все элементы. Алгоритм можно конечно оптимизировать за счет дополнительных операций, но худший сценарий выполнения они обычно не меняют.

А когда дополнительные операции влияют?

Представим, что мы хотим получить суммы всех элементов в списке. Для этого пройдемся по всем элементам списка, а внутри запустим еще цикл, который сложит эти элементы с элементом из внешнего цикла:

```python
nums = [3, 4, 5, 6, 8, 9, 7, 4, 1]

sums = []

for num in nums:
    for num2 in nums:
        sums.append(num + num2)
print(sums)
```

В данном случае мы получаем сложность **O(n\*\*2)** - квадратичную сложность алгоритма: количество операций является квадратом от количества элементов в худшем случае.

Бывает сложность алгоритма, которая считается константной - например, получение элемента списка по индексу. Такая сложность записывается как **O(1)**.

### Сложность по памяти

В некоторых случаях у нас узким местом является время выполнения программы, в других - память, которая потребуется для ее работы. Причем имеется в виду память дополнительная, которая потребуется именно алгоритму - входные данные мы и так уже учли.

Например, для задач с накопителями, счетчиками, минимальными/максимальными значениями нужна одна переменная. Мы можем сказать, что для таких алгоритмов требуется константная дополнительная память - **O(1)**.

В случае фильтрации нам уже потребуется дополнительный список для элементов, а значит мы оценим алгоритм по памяти как **O(n)**.

Вспомним сортировку списков. Функция `sorted()` вернет новый список - **O(n)**:

```python
nums = [5, 3, 2, 1, 4]
nums2 = sorted(nums)  # Создает новый список
print(id(nums))
print(id(nums2))
nums.sort()  # Отсортирует оригинальный список
print(nums)
```

Метод `.sort()` же поменяет значения в самом списке, значит требуется **O(1)** памяти.

При этом оба способа сортировки одинаковы с точки зрения сложности по времени.

## Цикл while

Для алгоритмов с линейной сложностью отлично подходит цикл `for` - просто перебираем элементы последовательности до конца или до случая, когда вызовем `break`.

Для более сложных алгоритмов нам пригодится цикл `while`.

На простом языке его логику можно описать так:

> До тех пор, пока не выполнится условие: выполни вот это действие

Как можно напечатать числа от 0 до 9? В случае цикла `for`:

```python
for i in range(10):
    print(i)
```

Что можно выразить через цикл `while`:

```python
i = 0
while i < 10:
    print(i)
    i += 1
```

Если мы не хотим, чтобы наш цикл работал бесконечно, нам важно, чтобы значение переменных, с помощью которых считалось условия цикла, менялось.

`while` полезнее `for`, когда у нас нет последовательности, которую можно перебрать, например все числа.

Возьмем все числа больше 10, возведем в квадрат, и суммируем цифры получившегося числа. Отберем 5 таких чисел, сумма цифр которых будет меньше 10:

```python
i = 11  #  Первое число
numbers = []
while len(numbers) < 5: # Собираем до 5 чисел
    square = i ** 2  # Возводим в квадрат
    square = str(square)  # Превращаем квадрат в строку
    digits = [int(digit) for digit in square]  # Перебираем все цифры в строке
    if sum(digits) < 3:  # Число подходит
        numbers.append(int(square))
    i += 1  # Переходим к следующему числу
print(numbers)
```

Иногда мы хотим, чтобы цикл был бесконечным. Тогда для выхода из него используется `break`.

Отличный пример - интерактивные программы:

```python
while True:  # условие ВСЕГДА удовлетворено
    s = input("Введите слово, чтобы напечатать (или ничего для конца программы): ")
    if not s:
        break
    print(s)
```

### Угадай число

Напишем программу:

Компьютер загадывает случайное число от 1 до 16, а мы попробуем его угадать.

В Python есть модуль `random`, позволяющий случайно (почти) выбрать некий элемент.

```python
import random
```

Загадаем случайное число от 1 до 16:

```python
random.randint(1, 16)
```

Можно загадать число из `range()`, а значит добавить шаг:

```python
random.randrange(1, 20, 2)
```

Или выбрать случайный элемент из последовательности:

```python
random.choice(["abc", "def", "qwe", "rty"])
```

Итак, реализуем программу:

```python
guessed_num = random.randint(1, 16)

while True:
    guess = int(input("Введите число от 1 до 16 включительно: "))
    if guess == guessed_num:
        print("Угадали!")
        break
```

Побудем на месте компьютера и попробуем сами решить алгоритм поиска числа.

Начнем с 1, будем добавлять по 1, в худшем случае это займет у нас 16 попыток.

Сложность нашего алгоритма "угадайки" - **O(n)**.

Что, если программа может давать подсказки?

```python
guessed_num = random.randint(1, 16)

while True:
    guess = int(input("Введите число от 1 до 16 включительно: "))
    if guess == guessed_num:
        print("Угадали!")
        break
    elif guess > guessed_num:
        print("Число больше, чем загаданное")
    else:
        print("Число меньше, чем загаданное")
```

В таком случае мы можем действовать по следующему алгоритму:

* попробуем число посередине между 1 и 16 ( скажем 8)
* если число больше - берем середину между 1 и 7
* если меньше - от 9 до 16
* повторяем, пока не угадаем

Теперь в худшем случае мы будем знать ответ после 4 попыток!

Этот алгоритм называется **бинарным поиском**.


### Бинарный поиск


На каждом витке поиска количество значений, которые мы проверяем, сокращается в 2 раза, пока мы не найдем нужное.

В нашей задаче 16 -> 8 -> 4 -> 2 -> 1 число.

Если бы чисел было 32, количество попыток выросло бы всего на 1, 64 - на 2 и т.д.

Таким образом, сложность бинарного поиска - *логарифмическая*, или **O(log n)** (логарифм по основанию 2).

Как реализовать бинарный поиск в коде c циклом `while`? Поищем случайное в списке:

```python
numbers = list(range(1, 17))
number = random.randrange(1, 17)

left = 0
right = len(numbers) - 1

while left <= right:
    middle = (left + right) // 2
    if numbers[middle] == number:
        print(f"Нашли! Число {numbers[middle]}")
        break
    elif numbers[middle] > number:
        right = middle - 1
    else:
        left = middle + 1
```

Вот его визуализация:

```python
from IPython.display import clear_output                      # эта часть нужна для
import time                                                   # визуализации алгоритмов

numbers = list(range(1, 17))
number = random.randrange(1, 17)

left = 0
right = len(numbers) - 1

while left <= right:
    middle = (left + right) // 2

    clear_output(wait=True)                                   # эта
    print(" ".join(str(n).ljust(2) for n in numbers))         # часть
    print("   " * left + "^" + str(numbers[left]))            # нужна
    print("   " * right + "^" + str(numbers[right]))          # для
    time.sleep(2)                                             # визуализации
    print("   " * middle + "*")                               # алгоритма

    if numbers[middle] == number:
        print(f"Нашли! Число {numbers[middle]}")
        break
    elif numbers[middle] > number:
        right = middle - 1

        print("Не нашли!")                                    # эта часть
        time.sleep(2)                                         # нужна для
        print(f"Сдвигаем правую границу к {numbers[right]}")  # визуализации
        time.sleep(5)                                         # алгоритма
    else:
        left = middle + 1

        print("Не нашли!")                                    # эта часть
        time.sleep(2)                                         # нужна для
        print(f"Сдвигаем левую границу к {numbers[left]}")    # визуализации
        time.sleep(5)                                         # алгоритма
```

Обратите внимание - бинарный поиск не имел бы смысла, если бы мы искали в не отсортированной последовательности! Мы бы не были уверены, что все элементы до нашего индекса меньше, а после - больше.

*Бонус*. В Python есть модуль `bisect`, который реализует бинарный поиск. В частности, функция `bisect()` ищет индекс для вставки элемента в список - все элементы слева от найденного индекса меньше либо равны нашему элементу:

```python
import bisect

nums = [1, 2, 5, 7, 11, 21, 21, 25, 26, 30]
print(bisect.bisect(nums, 23))
```

Можно посмотреть на [исходный код](https://github.com/python/cpython/blob/3.14/Lib/bisect.py#L40) и проверить, как реализован бинарный поиск в Python.

А функция `insort()` сначала найдет, а затем вставит на нужное место элемент:

```python
bisect.insort(nums, 23)
print(nums)
```

Сложность `insort()` в любом случае *O(n)*: такова сложность добавления элемента в список.

Сортировка в Python имеет сложность **O(n \* log n)**, это довольно ресурсоемкая операция. Поэтому если у нас уже отсортирован список, лучше добавлять значения через `insort()`.

## Алгоримтические паттерны

К счастью нам чаще всего не нужно самому придумывать решения: для многих алгоритмических задач уже придуманы решения в зависимости от их паттернов.

### Два указателя

Под указателями имеется в виду индексы элементов. Вместо одного элемента мы будем сравнивать сразу два (*two pointers*).

Вспомним задачу про палиндром:

```python
s = "А роза упала на лапу Азора!"
letters = ""
for char in s:
    if char.isalpha():
        letters += char.lower()
if letters == letters[::-1]:
    print("Палиндром!")
else:
    print("Не палиндром!")
```

Сложность алгоритма - **O(2n)**, округляем до **O(n)**. Можем решить задачу с той же сложностью, но ускорив в 2 раза и без дополнительной памяти в виде `letters`.

Скажем, мы можем двигаться с двух сторон последовательности `s`:

* первый индекс - `0`, последний - `len(s) - 1`
* запускаем цикл, пока оба индекса не сойдутся в одной точке
* проверяем элементы по индексам: если один из них чем-то не подходит - перейдем к следующему
* если оба заканчивают задачу - `break`

```python
s = "А роза упала на лапу Азора!"

left = 0
right = len(s) - 1

while left <= right:
    if not s[left].isalpha():
        left += 1
        continue
    if not s[right].isalpha():
        right -= 1
        continue
    if s[left].lower() != s[right].lower():
        print("Не палиндром!")
        break
    left += 1
    right -= 1
print("Палиндром!")
```

Вот его визуализация:

```python
s = "А роза упала на лапу Азора!"

left = 0
right = len(s) - 1

while left <= right:
    clear_output(wait=True)                                # эта часть
    print(s)                                               # нужна для
    print(" " * left + "^" + s[left])                      # визуализации
    print(" " * right + "^" + s[right])                    # алгоритма

    if not s[left].isalpha():
        print(f"Пропускаем {s[left]!r}")                   # эта часть
        time.sleep(3)                                      # нужна для
        print("Двигаем левый указатель")                   # визуализации
        time.sleep(3)                                      # алгоритма

        left += 1
        continue
    if not s[right].isalpha():
        print(f"Пропускаем {s[right]!r}")                  # эта часть
        time.sleep(3)                                      # нужна для
        print("Двигаем правый указатель")                  # визуализации
        time.sleep(3)                                      # алгоритма

        right -= 1
        continue
    if s[left].lower() != s[right].lower():
        print("Не палиндром!")
        break

    print(f"Символы {s[left]!r} и {s[right]!r} совпадают")  # эта часть
    time.sleep(3)                                           # нужна для
    print("Двигаем оба указателя")                          # визуализации
    time.sleep(3)                                           # алгоритма

    left += 1
    right -= 1
print("Палиндром!")
```

Указатели могут смотреть на 2 разные последовательности. Обычно они должны быть отсортированы, мы сравниваем порядок элементов в них и проверяем повторы/расхождения/пересечения.

Рассмотрим задачу. Даны два отсортированных списка. Найдем первый общий элемент:

```python
l1 = [1, 3, 5, 7]
l2 = [2, 4, 7, 9]

i = j = 0

while i < len(l1) and j < len(l2):
    if l1[i] == l2[j]:
        print(f"Найден повтор - {l1[i]}")
        break
    elif l1[i] < l2[j]:
        i += 1
    else:
        j += 1
```

Вот его визуализация:

```python
l1 = [1, 3, 5, 7]
l2 = [2, 4, 7, 9]

i = j = 0

while i < len(l1) and j < len(l2):
    clear_output(wait=True)                                   # эта
    print(" ".join(str(n).ljust(2) for n in l1))              # часть
    print("   " * i + "^" + str(l1[i]))                       # нужна для
    print(" ".join(str(n).ljust(2) for n in l2))              # визуализации
    print("   " * j + "^" + str(l2[j]))                       # алгоритма

    if l1[i] == l2[j]:
        print(f"Найден повтор - {l1[i]}")
        break
    elif l1[i] < l2[j]:
        i += 1

        print("Двигаем меньший элемент")                      # эта часть нужна для
        time.sleep(2)                                         # визуализации алгоритмов
    else:
        j += 1

        print("Двигаем меньший элемент")                      # эта часть нужна для
        time.sleep(2)                                         # визуализации алгоритмов
```

Алгоритм:

* оба индекса начинаются с `0`
* цикл работает, пока один из индексов не достигнет конца своей последовательности
* если элементы сходятся - двигаем оба индекса
* если нет - двигаем так, чтобы элементы стали ближе (например, меньший к большему)

Найдем элементы первого отсортированного списка, которых нет во втором:

```python
l1 = [1, 2, 3, 5, 7, 8, 10]
l2 = [2, 4, 5, 8]

i = j = 0
missing = []

while i < len(l1) and j < len(l2):
    if l1[i] == l2[j]:
        i += 1
        j += 1
    elif l1[i] < l2[j]:
        missing.append(l1[i])
        i += 1
    else:
        j += 1

assert missing == [1, 3, 7, 10], missing
```

Вот его визуализация:

### Скользящее окно

Это более продвинутая версия "двух указателей", где нужно проанализировать элементы между этими указателями. Указатели создают рамку - окно (*sliding window*), которое соответствует условиям задачи и которое мы постепенно сдвигаем вправо.

Описание алгоритма:

* оба указателя становятся в начало последовательности
* правый указатель движется, пока окно не удовлетворит значение
* фиксируем промежуточный результат для окна
* сдвигаем левый указатель, чтобы условие опять было не удовлетворено
* опять двигаем правый указатель до нового окна
* и так до конца последовательности

Посмотрим задачу.

Дан список чисел. Надо найти суммы всех окон длины `k` в списке.

Мы могли бы решить задачу вложенным циклом и получить **O(n \*\* 2)** сложность:

```python
l = [2, 5, 1, 4, 3, 9, 7, 2, 6, 5, 8]
k = int(input("Введите k: "))
result = []

for i, num in l:
    total = num
    for j in range(i + 1, i + k):
        if j == len(l):
            break
        total += l[j]
    result.append(total)
```

Со скользящим окном мы получим **O(n)**:

```python
l = [2, 5, 1, 4, 3, 9, 7, 2, 6, 5, 8]
k = int(input("Введите k: "))
result = []

left, right = 0, 0
total = 0

while right < len(l):  # Движемся до конца списка
    if right - left < k:  # Окно меньше нашего лимита
        total += l[right]  # Увеличиваем значение
        right += 1  # И расширяем окно вправо
    else:  # Окно достигло лимита
        result.append(total)  # Фиксируем результат
        total -= l[left]  #  Уменьшаем его на значение слева
        left += 1  # И сдвигаем левую границу вправо
print(result)
```

Вот его визуализация:

```python
l = [2, 5, 1, 4, 3, 9, 7, 2, 6, 5, 8]
k = int(input("Введите k: "))
result = []

left, right = 0, 0
total = 0

while right < len(l):  # Движемся до конца списка
    clear_output(wait=True)                                   # эта часть
    print(" ".join(str(n).ljust(2) for n in l))               # нужна для
    print("   " * left + "^" + str(l[left]))                  # визуализации
    print("   " * right + "^" + str(l[right]))                # алгоритма
    time.sleep(3)

    if right - left < k:  # Окно меньше нашего лимита
        total += l[right]  # Увеличиваем значение
        right += 1  # И расширяем окно вправо

        print(f"Длина окна {right - left}, сумма - {total}")  # эта часть
        print("Сдвигаем правую границу")                      # нужна для
        time.sleep(3)                                         # визуализации алгоритма
    else:  # Окно достигло лимита
        print("Фиксируем сумму предыдущего окна")             # эта часть нужна для
        time.sleep(1.5)                                       # визуализации алгоритма

        result.append(total)  # Фиксируем результат
        total -= l[left]  #  Уменьшаем его на значение слева
        left += 1  # И сдвигаем левую границу вправо

        print("Сдвигаем левую границу")                       # эта часть нужна для
        time.sleep(1.5)                                       # визуализации алгоритма
print(result)
```

В предыдущих примерах окна пересекались, но мы можем искать и непересекающиеся окна. В таком случае мы не постепенно двигаем левый указатель, а сразу перемещаем его на нужное место после условия "разрыва" окна.

Найдем суммы окон, в которых нет `0`:

```python
l = [2, 5, 0, 4, 0, 0, 7, 2, 0, 5, 8]
result = []
left = right = 0
total = 0
while right < len(l):
    if not l[right]:  # Нашли 0
        if total:  # Если есть что сохранить
            result.append(total)  # Сохраняем
            total = 0  # И обнуляем
        left = right + 1  # Передвигаем левую границу на следующий элемент после 0
    total += l[right]  # Всегда увеличиваем промежуточную сумму
    right += 1  # Всегда сдвигаем границу окна вправо
if total:  # Не забываем про последнее окно
    result.append(total)
print(result)
```

## Подсчет со словарем

Словарь - очень быстрая структура данных. Мы можем ориентироваться на сложность *O(1)* при получении элемента по ключу или проверки, есть ли в словаре элемент. Поэтому часто его используют в качестве счетчика.

[Здесь](https://wiki.python.org/moin/TimeComplexity) можно ознакомиться со сложностью алгоритмов всех коллекций в Python.

Алгоритм:

* берем элемент последовательности
* если его нет в словаре - добавляем со значением 1
* если есть - увеличиваем на 1

Рассмотрим задачу.

Найдем самые повторяющиеся символы в строке:

```python
s = "abracadabra"
d = {}  # Здесь мы собираем частоты символов
max_count = 0  # А здесь храним максимальное значение
for char in s:
    if char not in d:
        d[char] = 1
    else:
        d[char] += 1
    max_count = max(max_count, d[char])  # Проверим, стало ли значение теперь максимальным

for char, count in d.items():
    if count == max_count:
        print(char)
```

Вот его визуализация:

```python
s = "abracadabra"
d = {}  # Здесь мы собираем частоты символов
max_count = 0  # А здесь храним максимальное значение
for char in s:
    clear_output(wait=True)                                   # эта часть
    print(f"Символ: {char}")                                  # нужна для
    time.sleep(3)                                             # визуализации алгоритма

    if char not in d:
        d[char] = 1
    else:
        d[char] += 1
    max_count = max(max_count, d[char])  # Проверим, стало ли значение теперь максимальным

    print(f"Счетчик: {d}")                                    # эта часть
    print(f"Максимальное значение: {max_count}")              # нужна для
    time.sleep(3)                                             # визуализации алгоритма

for char, count in d.items():
    if count == max_count:
        print(char)
```

В Python есть модуль `collections` с типом `defaultdict`, который позволяет упростить этот код:

```python
from collections import defaultdict
s = "abracadabra"
d = defaultdict(int)  # Если ключа нет - вернет минимальное число, то есть 0
max_count = 0
for char in s:
    d[char] += 1
    max_count = max(max_count, d[char])

for char, count in d.items():
    if count == max_count:
        print(char)
```

Найдем элементы, которые повторяются ровно один раз:

```python
l = [1, 2, 2, 3, 4, 4]
unique = set()
duplicates = set()
for num in l:
    if num in duplicates:
        continue
    if num in unique:
        unique.discard(num)
        duplicates.add(num)
    else:
        unique.add(num)
print(unique)
```

Паттерны можно комбинировать, например "скользящее окно" и частотный словарь/множество.

Найдем самое длинное окно без повторов:

```python
l = [1, 2, 2, 3, 4, 4]

longest = 0  # Инициализируем максимальное значение
window = set()  # Храним уникальные элементы окна
left = right = 0  # Устанавливаем указатели окна
while right < len(l):
    if l[right] in window:  # Нашли повтор
        longest = max(longest, len(window)) # Фиксируем (возможно) новый максимум
        while l[right] in window:  # Пока повтор есть в окне
            window.discard(l[left]) # Удаляем из окна все элементы слева
            left += 1  # И двигаем окно влево
    window.add(l[right])  # Наконец добавляем элемент в окно
    right += 1  # И двигаем правую границу

longest = max(longest, len(window))  #  Не забываем про последнее окно!
print(longest)
```

Вот его визуализация:

```python
l = [1, 2, 2, 3, 4, 4]

longest = 0  # Инициализируем максимальное значение
window = set()  # Храним уникальные элементы окна
left = right = 0  # Устанавливаем указатели окна
while right < len(l):
    clear_output(wait=True)                                   # эта
    print(" ".join(str(n).ljust(2) for n in l))               # часть
    print("   " * left + "^" + str(l[left]))                  # нужна для
    print("   " * right + "^" + str(l[right]))                # визуализации
    time.sleep(3)                                             # алгоритма

    if l[right] in window:  # Нашли повтор
        print("Найден повтор, пытаемся удалить его из окна")  # эта часть нужна
        time.sleep(3)                                         # для визуализации алгоритма

        longest = max(longest, len(window)) # Фиксируем (возможно) новый максимум
        while l[right] in window:  # Пока повтор есть в окне
            window.discard(l[left]) # Удаляем из окна все элементы слева
            left += 1  # И двигаем окно влево

            clear_output(wait=True)                           # эта
            print(" ".join(str(n).ljust(2) for n in l))       # часть
            print("   " * left + "^" + str(l[left]))          # нужна для
            print("   " * right + "^" + str(l[right]))        # визуализации
            time.sleep(3)                                     # алгоритма

    window.add(l[right])  # Наконец добавляем элемент в окно
    right += 1  # И двигаем правую границу

    print("Уникальные элементы")                              # эта часть
    print(" ".join(str(n).ljust(2) for n in window))          # нужна для
    time.sleep(3)                                             # визуализации алгоритма

longest = max(longest, len(window))  #  Не забываем про последнее окно!
print(f"Самый длинный отрезок без повторов - {longest}")
```
